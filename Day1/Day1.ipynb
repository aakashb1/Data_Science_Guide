```python
#Q3) What is the survival rate by gender?
survived_male_rate = len(data[(data['Survived'] == 1) & (data['Sex'] == 'male')])/len(data)
print("Male survival rate is", survived_male_rate)

survived_female_rate = len(data[(data['Survived'] == 1) & (data['Sex'] == 'female')])/len(data)
print("Female survival rate is", survived_female_rate)
```

```python
#Q4) What is the average age of survivors vs non-survivors?
survivors = int(data[data['Survived'] == 1]['Age'].mean())
non_survivors = int(data[data['Survived'] == 0]['Age'].mean())
```
