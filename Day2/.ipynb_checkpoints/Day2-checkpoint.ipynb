```python
import pandas as pd
import os
import numpy as np
```

```python
data = pd.read_csv('.\Titanic-Dataset.csv')
data.head()
```

```python
#Q1) Grouped summary: survival rate by Pclass AND Sex combined
data.groupby(['Pclass', 'Sex']).agg({'Age':np.median}).reset_index()
```

```python
#Q3) Merge Titanic with a small manually created DataFrame (class descriptions)

class_descriptions = {"Pclass" : [1,2,3], "class_description" : ['Upper class', 'Middle Class', 'Lower class']}
class_descriptions = pd.DataFrame(data = class_descriptions)
class_descriptions.head()

data.merge(class_descriptions, how = 'inner')
```

```python
#Q3) Use apply to create an age bucket column: child / adult / senior
def classify(age):
    if age < 21:
        return 'Child'
    elif 21 < age < 60:
        return 'Adult'
    elif age > 60:
        return 'Senior'
    else:
        return None

data['Age Bucket'] = data['Age'].apply(classify)
```

```python
#Q4) Create a pivot table: rows = Pclass, columns = Sex, values = mean Fare
table = pd.pivot_table(data, values = "Fare", index = ['Pclass'], columns = ['Sex'], aggfunc = "mean")
```
