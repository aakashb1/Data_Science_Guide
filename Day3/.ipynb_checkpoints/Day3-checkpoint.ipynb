```python
import pandas as pd
import os
import numpy as np
```

```python
data = pd.read_csv('AB_NYC_2019.csv')
```

```python
#Q1) Fix the price column: remove $ and commas, cast to float

#data['price'] = data['price'].apply(lambda x : x.astype(str).str.replace('$', ''))
data['price'] = data['price'].replace('$','', regex=True)
data['price'] = data['price'].replace(',','', regex=True)

data['price'] = data['price'].apply(float)
```

```python
#Q2) Find and remove duplicate rows
print(np.sum(data.duplicated()), 'values are duplicated')
```

```python
#Q3) Detect outliers in price using IQR — flag them, do not remove yet
print(data['price'].isna().sum(), 'null values found')

p_25, p_75 = np.percentile(data['price'], 25), np.percentile(data['price'], 75)

IQR = p_75 - p_25
cut_off = 1.5 * IQR
lower_bound = p_25 - cut_off
upper_bound = p_75 + cut_off

def classify(price):
    if price < lower_bound:
        return True
    elif price > upper_bound:
        return True
    else:
        return False

data['is_outlier'] = data['price'].apply(classify)
data.head()
```
